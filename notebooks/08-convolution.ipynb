# Convolution and gradient comparison

**Objective:** Understand and execute a small neural-network computation.

Valid cross-correlation applies a shared kernel across spatial locations.

Synthetic data only; see research/protocol.md and limitations.md.

```python
import numpy as np
from dl_from_scratch import *
import torch
r=np.random.default_rng(11); x=r.normal(size=(2,1,5,5)); l=Conv2D(1,2,3,r); z=l.forward(x); g=np.ones_like(z); dx=l.backward(g)
a=torch.tensor(x,requires_grad=True); w=torch.tensor(l.w,requires_grad=True); b=torch.tensor(l.b,requires_grad=True); ref=torch.nn.functional.conv2d(a,w,b); ref.sum().backward()
print(np.max(abs(z-ref.detach().numpy())),np.max(abs(dx-a.grad.numpy())))
```
